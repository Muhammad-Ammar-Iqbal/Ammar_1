# Q12

```python
# for loop
guests = []
count = 0
for i in range(100):
    name = input("Enter guest name: ")

    if name in guests:
        print("This name is already entered. Enter another name.")
        
    else:
        guests.append(name)
        count +=1
        print("Guest added.")
    if count == 10:
     break
print("Guest list:", guests)
```

```python
# While loop
guests = []
i = 0
while True:
    name = input("Enter the name of guest : ")
    if name in guests:
        print("The guest is already entered : ")
        continue
    else:
        guests.append(name)
        i +=1
    if i == 10:
     break
print("Guests name :",guests)
    
    
```

# Q14

```python
menu = []
discard_dishes = ["fish","qeema","nihari"]
count = 0
for i in range(100):
    dish = input("Enter the dish : ").lower()
    if dish in discard_dishes:
        print("This dish is not available.")
        continue
    else:
        menu.append(dish)
        print("Dish added in menu")
        count +=1
    if count == 5:
        break
dish_prices = {}
total_dishes_price = 0
menu_price_per_head = {}
total_per_head_cost = 0
for i in (menu):
    price = float(input(f"Give me the total price of 300 people for this dish : {i}"))
    dish_prices[i] = price
    total_dishes_price += price
    per_head_cost = price / 300
    total_per_head_cost += per_head_cost
    menu_price_per_head[i] = per_head_cost
    print(f"The per head cost for {i} is : {per_head_cost}" )

print("Total menu price:", total_dishes_price)
print("Per head cost of complete menu:", total_per_head_cost)
print("Menu prices:", dish_prices)
print("Per head prices:", menu_price_per_head)



 
```
