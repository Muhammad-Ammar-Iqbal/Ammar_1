# Q3

```python
number_1 = float(input("Enter your first number"))
number_2 = float(input("Enter your second number"))
number_3 = float(input("Enter your third number"))
if (number_1 > number_2) and (number_1 > number_3) :
    print("First number is greater than the rest")
elif (number_2 > number_1) and (number_2 > number_3):
    print("Second number is greater than the rest")
elif (number_3 > number_1) and (number_3 > number_2):
    print("Third number is greater than the rest")
else:
    print("more than one numbers are equal")
```
